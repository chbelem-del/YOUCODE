Explication: On remarque qu'on a 505 lignes et 11 colonnes.
Toutes les colonnes sont de type objet (texte) apart l'age et depenses clients ( customer spendings).

```python
AVGAGE = df['Age'].mean()
AVGAGE
```

La moyenne d'age est de 46 ans.

```python
MEDAGE = df['Age'].median()
print(MEDAGE)
```

La medianne est de 46 ans, ce qui veut dire que 50% des clients ont un age au desssous de 46 ans.

On remarque egalement que la moyenne d'age est representatif de la population clientele et qu'il y'a une abscene des valeurs aberrantes.

```python
avgspending= df['Customer spendings'].mean()
medspending= df['Customer spendings'].median()
print(f"The average spending is {avgspending}, and the mediane spending is {medspending}")
```

```python
#La mediane d'age pour chaque pays:
df.groupby('Country')['Age'].median()
```

```python
df.columns
```

```python
#Graphe montrant les donées de Customer_spending par contry:
spendings= df.groupby("Country")['Customer spendings'].sum()
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))

spendings.plot(kind='bar')
plt.title("Customer_spending par contry")
plt.xlabel('Age')
plt.ylabel('spendings')
plt.xticks(rotation=90)

plt.show()
```

```python
#Supprimer les lignes ou le "customer spendings" est inferieur à 10:
df = df[df["Customer spendings"] >= 10]
df
```

```python
#Supprimer toute les lignes dupliquées:
df= df.drop_duplicates()
```

```python
#regrouper les colonnes choisis dans une seule dataframe:
df_final= df[["Age","Customer spendings","Gender","Country"]]
```

```python
#sauvegarder en CSV:
df_final.to_csv("Resultat.csv", index=False)
```
